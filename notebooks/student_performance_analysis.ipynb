# Student Performance Prediction — EDA & Preprocessing

This notebook performs exploratory data analysis and preprocessing on the Student Performance dataset:
1. Data Loading & Inspection
2. Data Quality Checks & Duplicate Removal
3. Median & Mode Missing Value Imputation
4. Outlier Analysis (IQR Method)
5. Feature Distributions & Bivariate Visualizations
6. Correlation Matrix Analysis
7. Cleaned Dataset Export

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="muted")
print("Libraries loaded successfully.")
```

## 1. Load Dataset

```python
raw_path = os.path.join('..', 'data', 'student_performance.csv')
df = pd.read_csv(raw_path)
print("Dataset Shape:", df.shape)
df.head()
```

```python
df.info()
df.describe(include='all')
```

## 2. Duplicate Removal & Quality Check

```python
print("Duplicate rows count:", df.duplicated().sum())
df.drop_duplicates(inplace=True)
df.reset_index(drop=True, inplace=True)
print("Shape after duplicate removal:", df.shape)
```

## 3. Missing Value Imputation (Median for numerical, Mode for categorical)

```python
print("Missing values before imputation:\n", df.isnull().sum())

num_cols = ['Study_Hours', 'Attendance', 'Previous_Marks', 'Assignment_Score', 'Internal_Marks', 'Sleep_Hours', 'Internet_Hours']
cat_cols = ['Family_Support', 'Extracurricular']

for col in num_cols:
    if df[col].isnull().sum() > 0:
        df[col].fillna(df[col].median(), inplace=True)

for col in cat_cols:
    if df[col].isnull().sum() > 0:
        df[col].fillna(df[col].mode()[0], inplace=True)

print("\nTotal Missing Values After Imputation:", df.isnull().sum().sum())
```

## 4. Outlier Analysis (IQR Method)

```python
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    outliers = df[(df[col] < Q1 - 1.5 * IQR) | (df[col] > Q3 + 1.5 * IQR)]
    print(f"{col} - Outliers count: {len(outliers)}")
```

## 5. Visualizations

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x='Performance', order=['Low', 'Medium', 'High'], palette='Set2')
plt.title('Target Variable: Performance Distribution')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
corr = df[num_cols].corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation Matrix Heatmap')
plt.show()
```

## 6. Save Cleaned Dataset

```python
clean_path = os.path.join('..', 'data', 'student_performance_cleaned.csv')
df.to_csv(clean_path, index=False)
print("Cleaned dataset saved successfully to:", clean_path)
```
